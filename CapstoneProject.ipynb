## 2. Problem Explanation 

Supermarket checkout lines frequently experience fluctuating customer volumes, which can lead to operational inefficiencies and long wait times if not managed properly. This project approaches this real-world challenge of managing customers waiting at a supermarket checkout counter, where customer are tracked by a Customer ID, Name, and Number of Items bought.

This problem really requires a data structure and algorithmic solution because customer service lines follow a strict First-In, First-Out (FIFO) order. A Queue data structure is practical for this scenario. This ensures fairness by serving the earliest customer first while dynamically handling continuous arrivals and departures without the rigidity of fixed-size arrays. Additionally, algorithmic approaches like linear search and traversal are essential to support critical management tasks—including tracking the number of waiting customers, locating specific individuals, removing canceled orders, and identifying the customer with the most items.

The intended purpose of the proposed Python program is to automate and streamline checkout queue management. By implementing a robust queue system with comprehensive operational capabilities, the solution ensures fairness, provides real-time visibility into queue status, reduces manual oversight, and improves the overall checkout efficiency for both customers and store staff.

## 4. Data Structure Implementation : Queue (FIFO)

Each customer is stored in a `CustomerNode`. The `CheckoutQueue` class keeps `front` and `rear`
references, so adding a customer is always at the rear and serving is always from the front.

```text
Front                                    Rear
 ↓                                        ↓
[Winnie] -> [Dennis] -> [Faith] -> [Ryan Ray] -> None
```

## 6. Class Definitions

The `CustomerNode` and `CheckoutQueue` classes below implement the FIFO queue and all its operations.

````python
class CustomerNode:
    def __init__(self, customer_id, name, num_items):
        self.customer_id = customer_id
        self.name = name
        self.num_items = num_items
        self.next = None


class CheckoutQueue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.size = 0

    # ---------------- ADD (Enqueue) ----------------
    def add_customer(self, customer_id, name, num_items):
        if self._find_node(customer_id):
            print(f"Error: Customer ID '{customer_id}' already exists.")
            return False
        new_node = CustomerNode(customer_id, name, num_items)
        if not self.front:
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node
        self.size += 1
        print(f"Customer '{name}' joined the checkout queue.")
        return True

    # ---------------- DISPLAY (Traversal) ----------------
    def display_queue(self):
        if not self.front:
            print("No customers waiting.")
            return
        print(f"{'ID':<6}{'Name':<18}{'Items':<6}")
        print("-" * 30)
        current = self.front
        while current:                     # Traversal algorithm
            print(f"{current.customer_id:<6}{current.name:<18}{current.num_items:<6}")
            current = current.next

    # ---------------- SERVE NEXT (Dequeue) ----------------
    def serve_next_customer(self):
        if not self.front:
            print("Error: No customers in the queue.")
            return None
        served = self.front
        self.front = self.front.next
        if not self.front:
            self.rear = None
        self.size -= 1
        print(f"Now serving: {served.name} (ID: {served.customer_id})")
        return served

    # ---------------- SEARCH (Linear Search) ----------------
    def _find_node(self, customer_id):
        current = self.front
        while current:
            if current.customer_id == customer_id:
                return current
            current = current.next
        return None

    def search_customer(self, customer_id):
        # Linear Search: O(n)
        node = self._find_node(customer_id)
        if node:
            print(f"Found: {node.customer_id} | {node.name} | {node.num_items} items")
            return node
        print(f"Customer ID '{customer_id}' not found.")
        return None

    # ---------------- REMOVE ----------------
    def remove_customer(self, customer_id):
        current = self.front
        prev = None
        if not current:
            print("Error: The queue is empty.")
            return False
        if current.customer_id == customer_id:
            self.front = current.next
            if not self.front:
                self.rear = None
            self.size -= 1
            print(f"Customer '{current.name}' removed from the queue.")
            return True
        while current and current.customer_id != customer_id:
            prev = current
            current = current.next
        if not current:
            print(f"Error: Customer ID '{customer_id}' not found.")
            return False
        prev.next = current.next
        if current == self.rear:
            self.rear = prev
        self.size -= 1
        print(f"Customer '{current.name}' removed from the queue.")
        return True

    # ---------------- COUNT ----------------
    def count_customers(self):
        return self.size

    # ---------------- CUSTOMER WITH MOST ITEMS ----------------
    def most_items(self):
        if not self.front:
            return None
        current = self.front
        most = current
        while current:
            if current.num_items > most.num_items:
                most = current
            current = current.next
        return most

    # ---------------- DISPLAY FIRST AND LAST ----------------
    def display_first_and_last(self):
        if not self.front:
            print("No customers waiting.")
            return None, None
        print(f"First: {self.front.name} (ID: {self.front.customer_id})")
        print(f"Last: {self.rear.name} (ID: {self.rear.customer_id})")
        return self.front, self.rear
````

## 7. Add Operation (Enqueue)

`add_customer()` attaches a new node after `rear` and moves `rear` forward — O(1), since no traversal
is needed.

````python
checkout = CheckoutQueue()
checkout.add_customer("C001", "Winnie Achieng", 12)
checkout.add_customer("C002", "Dennis Otieno", 3)
checkout.add_customer("C003", "Faith Wanjiku", 25)
checkout.add_customer("C001", "Duplicate Attempt", 5)  # should be rejected
checkout.display_queue()
````

## 8. Display Operation (Traversal)

`display_queue()` walks from `front` to the end, printing every waiting customer in arrival order.

````python
checkout.display_queue()
````

## 9. Search Operation — Linear Search

`search_customer()` checks each node's `customer_id` in sequence from the front — a **Linear
Search**, O(n) worst case.

````python
checkout.search_customer("C002")   # existing customer
checkout.search_customer("C999")   # non-existent customer
````

## 10. Serve Next Customer (Dequeue)

`serve_next_customer()` removes and returns the node at `front`, enforcing FIFO order.

````python
checkout.serve_next_customer()   # Winnie is served first, since she joined first
checkout.display_queue()
````

## 11. Remove Operation

`remove_customer()` lets a specific customer be removed from anywhere in the queue (e.g. if they leave the line), not just from the front.

````python
checkout.remove_customer("C003")
checkout.remove_customer("C999")   # error case: not found
checkout.display_queue()
````

## 13. Menu-Driven Program

The function below wraps every operation in a text menu. It is defined here but not auto-run (since
`input()` would pause the notebook) — call `menu()` in a new cell to try it interactively.

````python
def menu():
    checkout = CheckoutQueue()
    while True:
        print("""
--- SUPERMARKET CHECKOUT SYSTEM ---
1. Add customer to queue
2. Display queue
3. Serve next customer
4. Search for a customer
5. Remove a customer
6. Count customers waiting
7. Find customer with most items
8. Display first and last customer
9. Exit
""")
        choice = input("Enter choice: ").strip()
        if choice == "1":
            cid = input("Customer ID: ")
            name = input("Name: ")
            try:
                items = int(input("Number of items: "))
                checkout.add_customer(cid, name, items)
            except ValueError:
                print("Error: Number of items must be a number.")
        elif choice == "2":
            checkout.display_queue()
        elif choice == "3":
            checkout.serve_next_customer()
        elif choice == "4":
            checkout.search_customer(input("Enter Customer ID to search: "))
        elif choice == "5":
            checkout.remove_customer(input("Customer ID to remove: "))
        elif choice == "6":
            print(f"Customers waiting: {checkout.count_customers()}")
        elif choice == "7":
            m = checkout.most_items()
            print(f"Most items: {m.name} ({m.num_items})" if m else "No customers waiting.")
        elif choice == "8":
            checkout.display_first_and_last()
        elif choice == "9":
            print("Exiting...")
            break
        else:
            print("Invalid choice. Please try again.")
````

## 15. Complexity Analysis

| Operation | Time Complexity | Explanation |
|---|---|---|
| Add (enqueue) | O(1) | A `rear` pointer is kept, so a new node attaches directly with no traversal |
| Display (traversal) | O(n) | Visits every node once |
| Serve next (dequeue) | O(1) | Only the `front` pointer needs to move |
| Search (Linear Search) | O(n) | Worst case checks every node from the front |
| Remove (from anywhere) | O(n) | Must search for the node before re-linking |
| Count | O(1) | A running `size` counter is maintained |
| Most items | O(n) | Must inspect every node's item count |
| First and last | O(1) | Direct access via the `front` and `rear` pointers |

**Space Complexity:** O(n) for the queue itself; all operations use O(1) extra space.


## 16. Challenges

- **Keeping `front` and `rear` in sync:** every enqueue/dequeue must correctly update both pointers,
  including when the queue becomes empty or has exactly one node.
- **Removing from the middle:** the brief requires removing a specific customer, not just serving from
  the front, so the same "search with a trailing `prev` pointer" pattern from a general linked list was
  needed, with a check to fix `rear` if the removed node was the last one.
- **Finding "most items" without disrupting order:** unlike sorting, this needed a single traversal
  that tracks the maximum seen so far, leaving the FIFO order of the queue untouched.
